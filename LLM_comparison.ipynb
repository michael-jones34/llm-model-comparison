### Week 4 Exercise - Comparing how models assess fictional AI labs
The purpose of this notebook was to conduct a brief model comparison exercise. Four LLMs were tasked with assessing fictional AI labs in terms of AI safety risk and suitability as investment positions. The models evaluated were:
* GPT-4.1
* GPT-4.1 Mini
* Llama 3.1 8B
* Qwen 2.5 7B

The code below initializes the models, defines the system prompt, and creates a streaming Gradio interface used to generate the model outputs evaluated in this exercise.

```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
# Initialization

load_dotenv(override=True)

openai_api_key = os.getenv('OPENAI_API_KEY')
if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")

openai = OpenAI()

ollama_url = "http://localhost:11434/v1"
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
```

```python
# Models

MODEL_MAP = {
    "GPT-4.1": "gpt-4.1",
    "GPT-4.1 Mini": "gpt-4.1-mini",
    "Llama 3.1 8B": "llama3.1:8b",
    "Qwen 2.5 7B": "qwen2.5:7b"
}

clients = {"gpt-4.1": openai,
           "gpt-4.1-mini": openai,
           "llama3.1:8b": ollama,
           "qwen2.5:7b": ollama, }
```

```python
# System prompt

system_prompt = """
You are an analyst tasked with assessing AI labs as potential investment positions.
You must assess them for their AI safety risk as well as their suitability as an investment position based on likely ROI.
You should split your analysis into four sections as follows:
-Causal chain (3-6 bullets)
-Assumptions (3-6 bullets)
-ROI estimate (single short paragraph or 3 bullets)
-Summary (1-2 sentences)
Your report should be no longer than 200 words in markdown format.
"""
```

```python
def stream_answer(message, model_selector):
    model = MODEL_MAP[model_selector]

    api = clients[model]

    messages = [{"role": "system", "content": system_prompt},
                {"role": "user", "content": message},]

    stream = api.chat.completions.create(
        model=model,
        messages=messages,
        stream=True
    )
    result = ""
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        yield result
```

```python
message_input = gr.Textbox(label="Your message:", info="Enter a message for the LLM", lines=50)
model_selector = gr.Dropdown(["GPT-4.1", "GPT-4.1 Mini", "Llama 3.1 8B", "Qwen 2.5 7B"], label="Select model", value="GPT-4.1")
message_output = gr.Markdown(label="Response:")

view = gr.Interface(
    fn=stream_answer,
    title="Investment Position Assessments",
    inputs=[message_input, model_selector],
    outputs=[message_output],
    flagging_mode="never"
    )
view.launch()
```
